```python
import glob 
import os 
import shutil
```

```python
def merged_dataset(data_dir, output_file):
    txt_files = glob.glob(os.path.join(data_dir,"*.txt"))

    with open(output_file, "w", encoding='utf-8') as outfile:
        for file_path in txt_files:
            with open(file_path, 'r', encoding='utf-8') as infile:
                outfile.write(infile.read())
    print("Done merging ")
```

```python
# for train
data_dir = "../data/raw/train/train"
output_file = "../data/merged/train_merged.txt"
merged_dataset(data_dir, output_file)
```

```python
# for tokenizer 
data_dir = "../data/tokenizer"
output_file = "../data/tokenizer/train_tokenizer_corpus.txt"
merged_dataset(data_dir, output_file)

```

```python
# for train
data_dir = "../data/raw/valid/valid"
output_file = "../data/merged/validation_merged.txt"
merged_dataset(data_dir, output_file)
```
