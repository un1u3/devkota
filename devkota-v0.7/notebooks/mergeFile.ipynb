```python
# need to merge all files to merrged files
def merge_dataset(DATA_DIR, OUTPUT_FILE):

    # find all files ending in .txt format in that directory 
    txt_files = glob.glob(os.path.join(DATA_DIR, "*.txt"))
    with open(OUTPUT_FILE,"w", encoding = 'utf-8') as outfile:
        for file_path  in txt_files:
            if file_path == OUTPUT_FILE:
                continue
            
            with open(file_path, "r", encoding = 'utf-8') as infile:
                outfile.write(infile.read())
                # add new line after each 


```

```python
# for train 
merge_dataset('../data/raw/train/train',
              "../data/merged/train_merged.txt"
              )
# for test 
merge_dataset('../data/raw/valid/valid',
              "../data/merged/validation_merged.txt"
              )
```
