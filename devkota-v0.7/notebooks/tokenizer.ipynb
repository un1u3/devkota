```python
import sentencepiece as spm 
```

```python
spm.SentencePieceTrainer.train(
    input = '../data/tokenizer/train_tokenizer_corpus.txt',
    model_prefix = 'devkota_tokenizer',
    vocab_size = 8000,
    model_type = 'unigram',
    character_coverage = 0.9999,
    byte_fallback = True,
    normalization_rule_name = 'nfkc',
    pad_id = 0,
    unk_id = 1,
    bos_id = 2,
    eos_id = 3
)
```

```python
sp = spm.SentencePieceProcessor(
    model_file='../tokenizer/devkota_tokenizer.model'
)
```

```python
text = "नेपाली बाबु नेपालमा बनेको"
encoded_str  = sp.encode(text,out_type=str)
print(encoded_str)
```

```python
encoded_ids = sp.encode(text,out_type=int)
print(encoded_ids)
```

```python
print(f"from str :{sp.decode(encoded_str)}")
print(f"from ids :{sp.decode(encoded_ids)}")
```
